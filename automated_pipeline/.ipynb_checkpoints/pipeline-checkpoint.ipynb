# HCM Data Discrepancy Resolution Pipeline

**Workflow:**
1. Fill in credentials and select campaign / metric / scenario
2. Click **Run Pipeline** — Phase 0 (compare) always runs first
3. Based on scenario selection, Scenario A and/or B runs automatically
4. If any batches fail, use the **Retry** section at the bottom

| Scenario | Direction | API |
|----------|-----------|-----|
| **A** | Elastic → DB | `/project/task/v1/bulk/_create` |
| **B** | DB → Elastic | `/project/task/v1/bulk/_update` |

> **Note:** Credentials are never saved to disk. Re-enter them each session.

```python
import json, os, sys, base64, warnings
from pathlib import Path

import pandas as pd
import ipywidgets as widgets
from IPython.display import display, HTML

warnings.filterwarnings("ignore")

# Ensure pipeline/ is importable
repo_root = Path(".").resolve()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from pipeline import compare, verify, transform, ingest

with open("campaign_config.json") as f:
    CAMPAIGNS = json.load(f)
with open("metrics_config.json") as f:
    METRICS = json.load(f)

print("Setup complete.")
print("Campaigns :", list(CAMPAIGNS.keys()))
print("Metrics   :", list(METRICS.keys()))
```

```python
_W = {"description_width": "100px"}
_WIDE  = widgets.Layout(width="420px")
_MED   = widgets.Layout(width="300px")
_SHORT = widgets.Layout(width="180px")

# ── Database ──────────────────────────────────────────────
w_db_host = widgets.Text(description="DB Host:",     style=_W, layout=_WIDE,  placeholder="RDS endpoint")
w_db_port = widgets.IntText(value=5432, description="Port:", style=_W, layout=_SHORT)
w_db_name = widgets.Text(description="DB Name:",     style=_W, layout=_MED,  placeholder="ngcentralprd")
w_db_user = widgets.Text(description="DB User:",     style=_W, layout=_MED)
w_db_pass = widgets.Password(description="DB Pass:", style=_W, layout=_MED)

# ── Elasticsearch ─────────────────────────────────────────
w_es_user = widgets.Text(value="dst", description="ES User:", style=_W, layout=_MED)
w_es_pass = widgets.Password(description="ES Pass:",          style=_W, layout=_MED)

# ── OAuth token ───────────────────────────────────────────
w_token = widgets.Password(
    description="Auth Token:", style=_W,
    layout=widgets.Layout(width="600px"),
    placeholder="Paste OAuth token from cURL login"
)

# ── Campaign / Metric / Scenario ──────────────────────────
w_campaign = widgets.Dropdown(
    options=[(v["label"], k) for k, v in CAMPAIGNS.items()],
    description="Campaign:", style=_W, layout=_MED
)
w_metric = widgets.Dropdown(
    options=list(METRICS.keys()),
    description="Metric:", style=_W, layout=widgets.Layout(width="380px")
)
w_scenario = widgets.RadioButtons(
    options=[
        "Auto-detect (run whichever has gaps)",
        "Scenario A only  —  ES → DB  (Bulk Create)",
        "Scenario B only  —  DB → ES  (Bulk Update)",
        "Phase 0 only  —  Compare & report",
    ],
    description="Scenario:", style=_W
)
w_outdir = widgets.Text(
    value="output", description="Output dir:", style=_W, layout=_MED
)

# ── Layout ────────────────────────────────────────────────
display(widgets.HTML("<h3 style='margin-bottom:4px'>Database</h3>"))
display(widgets.HBox([w_db_host, w_db_port]))
display(widgets.HBox([w_db_name, w_db_user, w_db_pass]))

display(widgets.HTML("<h3 style='margin-bottom:4px'>Elasticsearch</h3>"))
display(widgets.HBox([w_es_user, w_es_pass]))

display(widgets.HTML("<h3 style='margin-bottom:4px'>API Auth Token</h3>"))
display(w_token)

display(widgets.HTML("<h3 style='margin-bottom:4px'>Run Settings</h3>"))
display(widgets.HBox([w_campaign, w_metric]))
display(w_scenario)
display(w_outdir)
```

```python
run_output = widgets.Output(layout=widgets.Layout(border="1px solid #ccc", padding="8px", max_height="600px", overflow_y="auto"))
run_btn    = widgets.Button(description="Run Pipeline", button_style="success", icon="play", layout=widgets.Layout(width="180px"))
status_lbl = widgets.HTML("")

# Stored results for retry / inspection
_state = {}

def on_run(b):
    run_btn.disabled = True
    status_lbl.value = "<b style='color:#1565C0'>Running...</b>"

    with run_output:
        run_output.clear_output()

        try:
            campaign_key = w_campaign.value
            campaign_cfg = CAMPAIGNS[campaign_key]
            metric_cfg   = METRICS[w_metric.value]
            scenario     = w_scenario.value
            output_dir   = w_outdir.value.strip() or "output"
            os.makedirs(output_dir, exist_ok=True)

            db_config = {
                "host":            w_db_host.value.strip(),
                "port":            w_db_port.value,
                "database":        w_db_name.value.strip(),
                "user":            w_db_user.value.strip(),
                "password":        w_db_pass.value,
                "sslmode":         "require",
                "connect_timeout": 30,
            }

            # ── PHASE 0: COMPARE ────────────────────────────────────
            print("=" * 60)
            print("PHASE 0 — COMPARE")
            print("=" * 60)

            result = compare.run_compare(
                campaign_cfg=campaign_cfg,
                metric_cfg=metric_cfg,
                db_config=db_config,
                es_username=w_es_user.value.strip(),
                es_password=w_es_pass.value,
                output_dir=output_dir,
            )
            _state["result"] = result
            _state["campaign_cfg"] = campaign_cfg

            missing_in_db       = result["missing_in_db"]       # Scenario A
            missing_in_elastic  = result["missing_in_elastic"]  # Scenario B

            if "Phase 0 only" in scenario:
                print("\nPhase 0 complete. Output files saved to:", output_dir)
                status_lbl.value = "<b style='color:green'>Compare done.</b>"
                return

            run_a = ("Auto" in scenario and len(missing_in_db) > 0) or "Scenario A" in scenario
            run_b = ("Auto" in scenario and len(missing_in_elastic) > 0) or "Scenario B" in scenario

            # ── SCENARIO A: ES → DB ─────────────────────────────────
            if run_a:
                print("\n" + "=" * 60)
                print("SCENARIO A — Re-ingest into DB (Bulk Create)")
                print(f"Candidates: {len(missing_in_db):,}")
                print("=" * 60)

                es_docs_df = result["es_docs_df"].copy()

                # Filter deleted
                del_col = "Data.isDeleted"
                if del_col in es_docs_df.columns:
                    before = len(es_docs_df)
                    es_docs_df = es_docs_df[es_docs_df[del_col] != True]
                    print(f"Filtered isDeleted: {before - len(es_docs_df):,} removed, {len(es_docs_df):,} remain")

                # Verify beneficiaries
                pb_col = "Data.projectBeneficiaryClientReferenceId"
                if pb_col in es_docs_df.columns and w_token.value:
                    pb_ids = es_docs_df[pb_col].dropna().tolist()
                    print(f"Verifying {len(pb_ids):,} beneficiary IDs...")
                    not_present = verify.verify_beneficiaries(
                        api_base=campaign_cfg["api_base"],
                        tenant_id=campaign_cfg["tenant_id"],
                        auth_token=w_token.value,
                        ids=pb_ids,
                    )
                    es_docs_df, removed = verify.segregate(es_docs_df, not_present, pb_col)
                    print(f"Segregated: {removed:,} removed, {len(es_docs_df):,} valid")

                # Get task IDs
                task_col = "Data.taskClientReferenceId"
                if task_col in es_docs_df.columns:
                    task_ids = es_docs_df[task_col].dropna().tolist()
                else:
                    task_ids = list(missing_in_db)
                print(f"Task IDs to transform: {len(task_ids):,}")

                # Fetch & transform from ES
                es_auth = "Basic " + base64.b64encode(
                    f"{w_es_user.value.strip()}:{w_es_pass.value}".encode()
                ).decode()
                tasks = transform.fetch_and_transform(
                    es_base=campaign_cfg["es_base_url"],
                    es_index=campaign_cfg["task"]["es_index"],
                    es_auth_header=es_auth,
                    task_ids=task_ids,
                )

                valid_tasks, issues = transform.validate_payload(tasks)
                if issues:
                    print(f"Validation: {len(issues):,} records dropped (null IDs)")
                print(f"Ready for ingestion: {len(valid_tasks):,}")

                payload_path = os.path.join(output_dir, f"scenario_a_payload_{result['ts']}.json")
                with open(payload_path, "w") as fp:
                    json.dump(valid_tasks, fp)
                print(f"Payload saved: {payload_path}")

                _state["scenario_a_tasks"] = valid_tasks

                # Bulk create
                api_url = campaign_cfg["api_base"] + campaign_cfg["task"]["api_create"]
                failed_path_a = os.path.join(output_dir, "failed_tasks_a.json")
                print(f"Calling: {api_url}")
                summary_a = ingest.bulk_ingest(
                    api_url=api_url,
                    auth_token=w_token.value,
                    tasks=valid_tasks,
                    failed_path=failed_path_a,
                )
                _state["failed_path_a"] = failed_path_a
                print(f"\nScenario A — Success: {summary_a['success']:,} | Failed: {summary_a['failed']:,}")

            # ── SCENARIO B: DB → ES ─────────────────────────────────
            if run_b:
                print("\n" + "=" * 60)
                print("SCENARIO B — Re-index into Elasticsearch (Bulk Update)")
                print(f"Candidates: {len(missing_in_elastic):,}")
                print("=" * 60)

                db_details_df = result["db_details_df"].copy()

                # Filter deleted
                if "isdeleted" in db_details_df.columns:
                    before = len(db_details_df)
                    db_details_df = db_details_df[db_details_df["isdeleted"] != True]
                    print(f"Filtered isDeleted: {before - len(db_details_df):,} removed")

                task_ids = db_details_df["clientreferenceid"].dropna().tolist()
                print(f"Fetching {len(task_ids):,} task objects from HCM API...")

                tasks = ingest.fetch_tasks_from_api(
                    api_base=campaign_cfg["api_base"],
                    tenant_id=campaign_cfg["tenant_id"],
                    auth_token=w_token.value,
                    client_reference_ids=task_ids,
                )
                print(f"Fetched: {len(tasks):,} tasks")

                payload_path = os.path.join(output_dir, f"scenario_b_payload_{result['ts']}.json")
                with open(payload_path, "w") as fp:
                    json.dump(tasks, fp)
                print(f"Payload saved: {payload_path}")

                _state["scenario_b_tasks"] = tasks

                api_url = campaign_cfg["api_base"] + campaign_cfg["task"]["api_update"]
                failed_path_b = os.path.join(output_dir, "failed_tasks_b.json")
                print(f"Calling: {api_url}")
                summary_b = ingest.bulk_ingest(
                    api_url=api_url,
                    auth_token=w_token.value,
                    tasks=tasks,
                    failed_path=failed_path_b,
                )
                _state["failed_path_b"] = failed_path_b
                print(f"\nScenario B — Success: {summary_b['success']:,} | Failed: {summary_b['failed']:,}")

            print("\n" + "=" * 60)
            print(f"Pipeline complete. Output saved to: {output_dir}")
            status_lbl.value = "<b style='color:green'>Pipeline complete.</b>"

        except Exception as exc:
            import traceback
            print(f"\nERROR: {exc}")
            traceback.print_exc()
            status_lbl.value = f"<b style='color:red'>Failed: {exc}</b>"
        finally:
            run_btn.disabled = False

run_btn.on_click(on_run)
display(widgets.HBox([run_btn, status_lbl]))
display(run_output)
```

## Retry Failed Batches
If `failed_tasks_a.json` or `failed_tasks_b.json` was created above, use the buttons below to retry.
Run until failures = 0.

```python
retry_output = widgets.Output(layout=widgets.Layout(border="1px solid #ccc", padding="8px"))
retry_a_btn  = widgets.Button(description="Retry Scenario A", button_style="warning", layout=widgets.Layout(width="180px"))
retry_b_btn  = widgets.Button(description="Retry Scenario B", button_style="warning", layout=widgets.Layout(width="180px"))

def _retry(scenario_label, failed_path_key, api_path_key):
    with retry_output:
        retry_output.clear_output()
        failed_path = _state.get(failed_path_key)
        if not failed_path or not os.path.exists(failed_path):
            print(f"No failed file found for {scenario_label}. Nothing to retry.")
            return
        campaign_cfg = _state.get("campaign_cfg")
        if not campaign_cfg:
            print("Run the pipeline first.")
            return
        api_url = campaign_cfg["api_base"] + campaign_cfg["task"][api_path_key]
        print(f"Retrying {scenario_label} from {failed_path}...")
        summary = ingest.retry_failed(
            failed_path=failed_path,
            api_url=api_url,
            auth_token=w_token.value,
        )
        print(f"Retry done — Success: {summary['success']:,} | Still failed: {summary['failed']:,}")

retry_a_btn.on_click(lambda b: _retry("Scenario A", "failed_path_a", "api_create"))
retry_b_btn.on_click(lambda b: _retry("Scenario B", "failed_path_b", "api_update"))

display(widgets.HBox([retry_a_btn, retry_b_btn]))
display(retry_output)
```
